# Part 1 — Everything is an object

Run:

````python
x = 42

type(x)
id(x)
repr(x)
````

Then:

````python
x = "hello"

type(x)
id(x)
repr(x)
````

### Questions

1. Did the type change?
2. Did the identity change?
3. What happened to the integer object?
4. Does the name `x` have a permanent type?

Write your answers in a Markdown cell.

### Reference answer

1. **Did the type change?** Yes. `x` first refers to an integer (`int`) and then to a string (`str`).
2. **Did the identity change?** Yes. Rebinding `x` makes it refer to a different object, so its `id()` changes.
3. **What happened to the integer object?** The name `x` stopped referring to it. If no other references to that integer object remain, Python is free to reclaim it. (For small integers, Python may internally reuse objects, so the exact `id()` behavior should not be treated as a memory-management experiment.)
4. **Does `x` have a permanent type?** No. The **object** has a type; the name `x` can be rebound to objects of different types.

# Part 3 — Mutation vs. rebinding

Compare:

````python
x = [1, 2, 3]
y = x
x.append(4)
````

with:

````python
x = [1, 2, 3]
y = x
x = [1, 2, 3, 4]
````

For each, inspect:

````python
x
y
x is y
````

### Question

What is the difference between `x.append(4)` and `x = [1, 2, 3, 4]`?

### Reference answer

`x.append(4)` **mutates the existing list in place**. If `y = x`, both names still refer to that same list, so both see the appended value and `x is y` remains `True`.

By contrast,

```python
x = [1, 2, 3, 4]
```

**rebinds** the name `x` to a new list object. `y` still refers to the original `[1, 2, 3]`, so `x is y` becomes `False`.

# Part 5 — A function that changes its input

Run:

````python
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
````

Then create a non-mutating version:

````python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
````

### Reflection

Which style would you prefer for a statistical function, and why?

### Reference answer

For most statistical/data-science functions, I would usually prefer the **non-mutating version** unless in-place modification is intentional and clearly documented.

Returning a new object makes the function easier to reason about because the caller's original data remain unchanged. This reduces unexpected side effects, makes debugging/testing easier, and is especially helpful when the same data object is used elsewhere in an analysis.

Mutation can still be appropriate when memory/performance matters, but it should be explicit.

# Part 6 — Shallow copying

Run:

````python
x = [[1, 2], [3, 4]]
y = x.copy()

print(x is y)
print(x[0] is y[0])
````

Then:

````python
x[0].append(99)

print(x)
print(y)
````

Explain why the inner list changed in both objects.

### Reference answer

`x.copy()` makes a **shallow copy**. The outer lists are different objects, so:

```python
x is y
```

is `False`.

However, the nested lists are still shared. Therefore:

```python
x[0] is y[0]
```

is `True`.

When `99` is appended to `x[0]`, the shared inner list changes, so the change appears inside both `x` and `y`.

Explain why a tuple could make conceptual sense for a coordinate pair.

### Reference answer

A latitude/longitude coordinate is naturally a **fixed pair of values**. A tuple communicates that the pair belongs together and is not intended to be modified element-by-element.

Why might `.get()` be preferable when a key may not exist?

### Reference answer

Dictionary indexing such as:

```python

```

raises a `KeyError` if the key does not exist.

By contrast:

```python

```

returns `None` by default, or a value you specify:

This is useful when a key is optional and its absence is expected rather than exceptional.

# Part 12 — Dynamic typing

Run:

````python
x = 10
print(type(x))

x = "ten"
print(type(x))

x = [10]
print(type(x))
````

Then:

````python
x = "10"
x + 5
````

Explain why Python can be dynamically typed while still having types.

### Reference answer

**Dynamically typed** does not mean “untyped.” Every Python object still has a type.

The dynamic part is that a **name is not permanently declared to have one type**. For example, `x` can first refer to an `int`, then a `str`, then a `list`.

The expression:

```python
"10" + 5
```

fails because Python still respects object types: it does not automatically combine a string and an integer using `+`.

# Part 13 — Type annotations

Write:

````python
def mean(values: list[float]) -> float:
    return sum(values) / len(values)
````

Run:

````python
mean([1.0, 2.0, 3.0])
mean.__annotations__
````

Then:

````python
def add(a: int, b: int) -> int:
    return a + b

add(2, 3)
add("hello", "world")
````

Explain what the annotations communicate and why annotations do not make Python statically typed.

### Reference answer

In:

```python
def mean(values: list[float]) -> float:
```

`list[float]` communicates that `values` is intended to be a list of floating-point values, and `-> float` communicates that the function is intended to return a float.

These are **annotations**, not automatic runtime restrictions. Likewise:

```python
def add(a: int, b: int) -> int:
    return a + b
```

can still be called with strings, and Python will concatenate them because the annotations are not enforced by the interpreter.

Annotations improve documentation, editor support, and static type checking, but they do not make Python statically typed.

# Part 14 — List comprehensions

Traditional loop:

````python
squares = []

for x in range(10):
    squares.append(x ** 2)
````

Rewrite:

````python
squares = [x ** 2 for x in range(10)]
````

Then:

````python
even_squares = [
    x ** 2
    for x in range(21)
    if x % 2 == 0
]
````

Write the equivalent traditional loop and compare readability.

### Reference solution

The traditional-loop equivalent is:

```python
even_squares = []

for x in range(21):
    if x % 2 == 0:
        even_squares.append(x ** 2)
```

For this simple transformation/filter, the comprehension is arguably clearer because the operation fits naturally in one readable expression:

```python
even_squares = [x ** 2 for x in range(21) if x % 2 == 0]
```

If the logic required several branches or intermediate calculations, the explicit loop would usually be easier to read.

# Part 18 — When not to use a comprehension

Consider:

````python
results = []

for person in people:
    if person["age"] >= 18:
        if person["status"] == "active":
            transformed = complicated_function(person)
            results.append(transformed)
````

Consider whether compressing this into one comprehension would actually improve the code.

### Principle

> **Use comprehensions when they make simple transformations clearer.**

Do not use them merely to minimize lines.

### Reference answer

Compressing this entire block into one comprehension would probably make it **harder to read**, not better.

There are two conditions plus a nontrivial transformation:

```python
if person["age"] >= 18:
    if person["status"] == "active":
        transformed = complicated_function(person)
```

A comprehension can technically express this, but the explicit loop makes the control flow and intermediate transformation easier to inspect and debug. The goal is readable code, not the fewest lines.

# Part 22 — Reflection

Answer these questions in a Markdown cell.

1. What is the relationship between a variable name and a Python object?
2. Why can mutability create unexpected behavior when passing objects to functions?
3. Why would you choose a dictionary instead of a list?
4. What does dynamically typed mean?
5. What problem can type annotations help solve?
6. When does a comprehension improve code?
7. Why should a statistician care about Python's object model and data structures?

### Reference answers

1. **What is the relationship between a variable name and a Python object?**  
   A name is a reference to an object. The object has a type, value, and identity; the name can later be rebound to another object.

2. **Why can mutability create unexpected behavior when passing objects to functions?**  
   A function may receive a reference to the same mutable object held by the caller. If the function mutates it, the caller sees the change too.

3. **Why choose a dictionary instead of a list?**  
   Use a dictionary when values are naturally accessed by meaningful keys (for example, subject ID → group) rather than primarily by position.

4. **What does dynamically typed mean?**  
   Names are not permanently declared with one type; they can refer to objects of different types at runtime.

5. **What problem can type annotations help solve?**  
   They communicate intended interfaces, improve editor/static-analysis support, and help catch type mismatches before runtime when used with a type checker.

6. **When does a comprehension improve code?**  
   When a simple transformation/filter can be expressed more directly and readably than an explicit loop.

7. **Why should a statistician care about Python's object model and data structures?**  
   Because aliasing, mutation, container choice, and data representation affect correctness, reproducibility, performance, and how safely statistical workflows manipulate data.
