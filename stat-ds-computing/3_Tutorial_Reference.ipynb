# Part 16 — Missing values

Create:

````python
missing_df = pd.DataFrame(
    {
        "id": [1, 2, 3, 4, 5],
        "age": [21, 32, None, 45, 28],
        "score": [81, None, 77, 92, 88]
    }
).set_index("id")
````

Inspect:

````python
missing_df
missing_df.isna()
missing_df.isna().sum()
````

Find missing scores:

````python
missing_df.loc[
    missing_df["score"].isna()
]
````

Find observed scores:

````python
missing_df.loc[
    missing_df["score"].notna()
]
````

### Reference answer

Use `df.isna().sum()`, `df.loc[df['score'].isna()]`, and `df.loc[df['score'].notna()]`.
