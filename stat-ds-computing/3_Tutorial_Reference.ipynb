2. What does the second argument control?

# Part 16 — Missing values

Create:

````python
missing_df = pd.DataFrame(
    {
        "id": [1, 2, 3, 4, 5],
        "age": [21, 32, None, 45, 28],
        "score": [81, None, 77, 92, 88]
    }
).set_index("id")
````

````python
missing_df
missing_df.isna()
missing_df.isna().sum()
````

Find missing scores:

````python
missing_df.loc[
    missing_df["score"].isna()
]
````

Find observed scores:

````python
missing_df.loc[
    missing_df["score"].notna()
]
````

### Reflection

Why is `score = 0` different from `score = missing` in statistical analysis?

### Reference answer

A score of `0` is an **observed value**. It means the score was measured and its value was zero.

A missing score means the value was **not observed or is unavailable**.

Treating missing values as zero would change the meaning of the data and could distort summaries such as means, distributions, and model estimates.

# Part 18 — Avoid chained assignment

Consider:

````python
df[df["score"] > 80]["score"] = 100
````

Prefer:

````python
df.loc[
    df["score"] > 80,
    "score"
] = 100
````

### Question

Why is the `.loc` version clearer? Connect your answer to Week 2's discussion of references and mutation.

### Reference answer

The `.loc` version is clearer because it performs the row selection and assignment in one explicit operation:

```python
df.loc[
    df["score"] > 80,
    "score"
] = 100
```

By contrast:

```python
df[df["score"] > 80]["score"] = 100
```

first creates an intermediate object and then tries to modify it. That intermediate object may be a copy rather than the original DataFrame.

This connects to the earlier discussion of references and mutation: it is important to know **which object is actually being modified**.

Write a reusable function:

````python
def select_high_scores(
    data: pd.DataFrame,
    minimum_score: float
) -> pd.DataFrame:
    ...
````

It should return observations whose scores are at least `minimum_score`.

Test:

and:

````python
select_high_scores(participants, 90)
````

### Extension

Modify the function so the caller can also specify which columns to return.

### Reference answer

A reusable function is:

```python
def select_high_scores(
    data: pd.DataFrame,
    minimum_score: float
) -> pd.DataFrame:
    return data.loc[
        data["score"] >= minimum_score
    ]
```

For the extension:

```python
def select_high_scores(
    data: pd.DataFrame,
    minimum_score: float,
    columns=None
) -> pd.DataFrame:
    mask = data["score"] >= minimum_score

    if columns is None:
        return data.loc[mask]

    return data.loc[
        mask,
        columns
    ]
```

Example:

```python
select_high_scores(
    participants,
    85,
    ["group", "score"]
)
```

# Part 24 — Reproducibility challenge

Imagine you wrote:

````python
good_data = participants.loc[
    participants["score"] > 80
]
````

Six months later, someone asks:

> Why did you choose 80?

Improve the workflow:

````python
MINIMUM_SCORE = 80

good_data = participants.loc[
    participants["score"] > MINIMUM_SCORE
]
````

### Discussion

Does naming the threshold improve reproducibility? What documentation is still missing?

### Reference answer

Naming the threshold improves reproducibility because the value is explicit and can be changed in one place:

```python
MINIMUM_SCORE = 80
```

However, the name alone does not explain **why 80 was chosen**.

Useful documentation might include:

- the scientific or practical rationale for the threshold;
- whether the rule should be `> 80` or `>= 80`;
- the source of the rule (protocol, policy, prior literature, etc.);
- when or why the threshold was changed.

Reproducibility requires both executable code and enough context to understand the analytical choices.

### Reference answer

### 1. DataFrames

A DataFrame provides labeled rows and columns plus tabular operations such as filtering, grouping, missing-value handling, joins, dtype management, and vectorized computation. A list of dictionaries can represent similar records, but does not provide these operations directly.

### 2. Series

```python
df["age"]
```

returns a one-dimensional Series.

```python
df[["age"]]
```

returns a one-column DataFrame.

### 3. `.loc` vs. `.iloc`

`.loc` selects using **labels**, while `.iloc` selects using **integer positions**.

### 4. Boolean filtering

```python
df["score"] >= 85
```

produces a Boolean Series containing one `True` or `False` value for each row.

### 5. Analysis populations

A filter is more than a programming operation because it defines which observations are included in the statistical analysis and therefore affects the scientific question being answered.

### 6. Validation

Code can run without errors while still implementing the wrong criterion. Inspecting the result helps detect logical errors.

### 7. Reproducibility

Explicit criteria make the inclusion/exclusion rules visible in code so another analyst can understand and recreate the same analysis population.
